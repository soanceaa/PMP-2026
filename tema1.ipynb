```python
a)N urmeaza o distributie geometrica deoarece probabilitatea sa obtinem stema este egala cu probabilitatea sa obtinem ban, ambele fiind egale cu 1/2. Pentru ca jocul sa se
termine la pasul n trebuie sa obtinem ban de n-1 ori iar apoi stema. Asadar P(N=n)=(1/2)^(n-1)*(1/2)=(1/2)^(n)
```

```python
import random

N = 0
S = 0

while True:
    N += 1
    moneda = random.choice(["s", "b"])

    if moneda == "s":
        break
    else:
        z = random.randint(1, 6)
        S += z - 3

print(N)
print(S)
```

```python
import random
import matplotlib.pyplot as plt

nr_simulari = 100000
S_valori = []

for i in range(nr_simulari):
    S = 0

    while True:
        moneda = random.choice(["s", "b"])

        if moneda == "s":
            break
        else:
            z = random.randint(1, 6)
            S += z - 3

    S_valori.append(S)

media_S = sum(S_valori) / nr_simulari

print(media_S)

plt.hist(S_valori, bins=range(min(S_valori), max(S_valori) + 2),
         edgecolor="black")
plt.xlabel("S")
plt.ylabel("Frecventa")
plt.title("Distributia lui S")
plt.show()
```

```python
import random
import matplotlib.pyplot as plt

nr_simulari = 10000

for p in [0.3, 0.7]:
    S_valori = []

    for i in range(nr_simulari):
        S = 0

        while True:
            moneda = random.choices(["s", "b"], [p, 1-p])[0]

            if moneda == "s":
                break
            else:
                z = random.randint(1, 6)
                S += z - 3

        S_valori.append(S)

    media_S = sum(S_valori) / nr_simulari

    print("p =", p)
    print("S =", media_S)

    plt.hist(S_valori, bins=range(min(S_valori), max(S_valori) + 2),
             edgecolor="black")
    plt.xlabel("S")
    plt.ylabel("Frecventa")
    plt.title("Distributia lui S pentru p = " + str(p))
    plt.show()
```

Exercitiul 3

Probabilităţile de preluare a unui client de către un anumit frizer sunt 3/13,6/13, respectiv 4/13 deoarece acestea sunt proporționale cu vitezele medii de lucru ale celor trei frizeri. Cum viteza totală este 3+6+4=13 clienți pe oră, probabilitățile ca un client să fie preluat de cei trei frizeri sunt 3/13,6/13, respectiv 4/13.

```python
import random
import numpy as np
import arviz as az
import matplotlib.pyplot as plt

nr_simulari = 10000
X_valori = []

for i in range(nr_simulari):

    frizer = random.choices(
        [1, 2, 3],
        weights=[3/13, 6/13, 4/13]
    )[0]

    if frizer == 1:
        X = np.random.exponential(scale=1/3)
    elif frizer == 2:
        X = np.random.exponential(scale=1/6)
    else:
        X = np.random.exponential(scale=1/4)

    X_valori.append(X)

X_valori = np.array(X_valori)

media = np.mean(X_valori)
deviatia_standard = np.std(X_valori)

print("Media lui X =", media)
print("Deviatia standard a lui X =", deviatia_standard)

az.plot_kde(X_valori)
plt.xlabel("Timpul de servire X (ore)")
plt.ylabel("Densitate")
plt.title("Densitatea aproximativa a lui X")
plt.show()
```
