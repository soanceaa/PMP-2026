Subpunctul a

```python
import numpy as np

stari = ["D", "M", "U"]
note = ["FB", "B", "S", "NS"]

pi = np.array([1/3, 1/3, 1/3])

A = np.array([
    [0.0, 0.5, 0.5],
    [0.5, 0.25, 0.25],
    [0.5, 0.25, 0.25]
])

B = np.array([
    [0.1, 0.2, 0.4, 0.3],
    [0.15, 0.25, 0.5, 0.1],
    [0.2, 0.3, 0.4, 0.1]
])

print("Distributia initiala pi:")
print(pi)

print("\nMatricea de tranzitie A:")
print(A)

print("\nMatricea de emisie B:")
print(B)

print("\nSuma probabilitatilor initiale:")
print(pi.sum())

print("\nSuma fiecarei linii din A:")
print(A.sum(axis=1))

print("\nSuma fiecarei linii din B:")
print(B.sum(axis=1))

print("\nVerificare:")
print("pi are suma 1:", np.isclose(pi.sum(), 1))
print("Fiecare linie din A are suma 1:", np.allclose(A.sum(axis=1), 1))
print("Fiecare linie din B are suma 1:", np.allclose(B.sum(axis=1), 1))
```

Subpunctul b

```python
prima_nota = "FB"
indice_nota = note.index(prima_nota)

delta_1 = pi * B[:, indice_nota]

print("Prima nota observata:", prima_nota)
print("\nCalculul valorilor delta_1:")

for j in range(len(stari)):
    print(f"delta_1({stari[j]}) = {pi[j]} * {B[j, indice_nota]} = {delta_1[j]}")

print("\nVectorul delta_1:")
print(delta_1)
```

Subpunctul c

```python
note_observate = ["FB", "FB", "S", "B", "B", "S", "B", "B", "NS", "B", "B", "S"]

coduri_note = {"FB": 0, "B": 1, "S": 2, "NS": 3}
observatii = np.array([coduri_note[n] for n in note_observate])

T = len(observatii)
N = len(stari)

delta = np.zeros((T, N))
psi = np.zeros((T, N), dtype=int)

delta[0] = delta_1

for t in range(1, T):
    for j in range(N):
        scoruri_precedente = delta[t - 1] * A[:, j]
        psi[t, j] = np.argmax(scoruri_precedente)
        delta[t, j] = B[j, observatii[t]] * np.max(scoruri_precedente)

print("Tabelul scorurilor delta:")
print("Moment | Dificil | Mediu | Usor")

for t in range(T):
    print(f"{t + 1:7} | {delta[t, 0]:.10f} | {delta[t, 1]:.10f} | {delta[t, 2]:.10f}")

print("\nTabelul starilor precedente psi:")
print("Moment | Dificil | Mediu | Usor")

for t in range(T):
    print(f"{t + 1:7} | {psi[t, 0]} | {psi[t, 1]} | {psi[t, 2]}")
```

Subpunctul d

```python
print("Valorile delta_1:")
for j in range(N):
    print(f"delta_1({stari[j]}) = {delta[0, j]:.10f}")

print("\nValorile delta_2:")
for j in range(N):
    print(f"delta_2({stari[j]}) = {delta[1, j]:.10f}")

print("\nVerificare manuala pentru delta_2(D):")
scoruri = delta[0] * A[:, 0]

for i in range(N):
    print(f"delta_1({stari[i]}) * a[{stari[i]},D] = {delta[0, i]:.10f} * {A[i, 0]:.2f} = {scoruri[i]:.10f}")

calcul = B[0, observatii[1]] * np.max(scoruri)

print("delta_2(D) calculat manual =", calcul)
print("delta_2(D) din tabel =", delta[1, 0])
```

Subpunctul e

```python
stare_finala = np.argmax(delta[T - 1])
scor_final = delta[T - 1, stare_finala]

secventa_optima = np.zeros(T, dtype=int)
secventa_optima[T - 1] = stare_finala

for t in range(T - 1, 0, -1):
    secventa_optima[t - 1] = psi[t, secventa_optima[t]]

nume_stari = ["Dificil", "Mediu", "Usor"]
secventa_rezultata = [nume_stari[i] for i in secventa_optima]

print("Secventa optima de dificultati:")

for t, stare in enumerate(secventa_rezultata, start=1):
    print(f"Testul {t}: {stare}")

print("\nScorul maxim al secventei:")
print(scor_final)

print("\nProbabilitatea comuna maxima:")
print(f"{scor_final:.15e}")
```

Subpunctul f

Algoritmul Viterbi ne ajută să determinăm cea mai probabilă secvență completă de dificultăți ale celor 12 teste, ținând cont de notele obținute și de probabilitățile inițiale, de tranziție și de emisie ale modelului. În cazul nostru, secvența obținută este D, U, D, U, D, M, D, U, D, U, D, M.
Este important de precizat că algoritmul nu alege pur și simplu dificultatea cu probabilitatea cea mai mare pentru fiecare test luat separat. El caută combinația de stări care are cea mai mare probabilitate pentru întreaga secvență de 12 teste. Din acest motiv, rezultatul depinde atât de notele observate, cât și de probabilitățile de trecere de la o stare la alta.
Scorul final obținut este aproximativ 2.109375 · 10⁻¹¹. Acesta reprezintă probabilitatea comună a secvenței optime de dificultăți și a notelor observate, nu probabilitatea condiționată a secvenței de dificultăți știind notele. Valoarea este mică deoarece se calculează probabilitatea comună pentru o secvență care conține 12 stări și 12 observații.
