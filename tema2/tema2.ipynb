```python

import numpy as np
import pandas as pd

np.random.seed(42)

n = 100_000

X = np.random.randint(1, 7, size=n)
Y = np.random.randint(1, 7, size=n)

suma = X + Y

A = suma >= 10
B = X % 2 == 0
C = (X % 2 == 0) | (Y % 2 == 0)
D = (X % 2 == 0) & (Y % 2 == 0)
E = suma == 8

p1 = np.mean(A[B])
p2 = np.mean(A[C])
p3 = np.mean(D[E])

pA = np.mean(A)
pB = np.mean(B)
pAB = np.mean(A & B)

rezultate = pd.DataFrame({
    "Subpunct": [
        "1. Suma >= 10 | X par",
        "2. Suma >= 10 | cel putin un zar par",
        "3. Ambele pare | suma = 8",
        "4. P(A intersectat B)",
        "4. P(A) * P(B)"
    ],
    "Estimare simulare": [
        p1, p2, p3, pAB, pA * pB
    ],
    "Valoare teoretica": [
        2/9, 5/27, 3/5, 1/9, 1/12
    ]
})

print(rezultate.round(4).to_string(index=False))

print("\nVerificarea independentei:")
print("P(A) estimat =", round(pA, 4))
print("P(B) estimat =", round(pB, 4))
print("P(A intersectat B) estimat =", round(pAB, 4))
print("P(A) * P(B) estimat =", round(pA * pB, 4))

if np.isclose(pAB, pA * pB, atol=0.002):
    print("Simularea sugereaza ca evenimentele sunt independente.")
else:
    print("Simularea sugereaza ca evenimentele nu sunt independente.")
```

```python

import pandas as pd

date = {
    "Facultate": ["Facultatea 1", "Facultatea 1",
                  "Facultatea 2", "Facultatea 2"],
    "Candidat": ["A", "B", "A", "B"],
    "Admisi": [72, 38, 4, 12],
    "Total": [80, 40, 40, 60]
}

df = pd.DataFrame(date)

df["Rata"] = df["Admisi"] / df["Total"] * 100

print("Ratele de admitere pe facultati:")
print(df)

global_data = df.groupby("Candidat")[["Admisi", "Total"]].sum()
global_data["Rata globala"] = (
    global_data["Admisi"] / global_data["Total"] * 100
)

print("\nRatele globale:")
print(global_data)
```
