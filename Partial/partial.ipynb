```python
import numpy as np
from hmmlearn.hmm import CategoricalHMM

hmm = CategoricalHMM(
    n_components=4,
    init_params=""
)

hmm.startprob_ = np.array([0.25, 0.25, 0.25, 0.25])

hmm.transmat_ = np.array([
    [0.2, 0.4, 0.0, 0.4],
    [0.4, 0.2, 0.4, 0.0],
    [0.0, 0.4, 0.2, 0.4],
    [0.4, 0.0, 0.4, 0.2]
])

hmm.emissionprob_ = np.array([
    [0.7, 0.2, 0.1],
    [0.2, 0.6, 0.2],
    [0.1, 0.3, 0.6],
    [0.5, 0.2, 0.3]
])

observatii = np.array([[0, 1, 2, 2, 0]])

log_prob, secventa_stari = hmm.decode(
    observatii,
    algorithm="viterbi"
)

probabilitate_comuna = np.exp(log_prob)

camere = secventa_stari + 1

print("Secventa cea mai probabila de camere:", camere)
print("Probabilitatea comuna:", probabilitate_comuna)
```

```python
hmm.transmat_ = np.array([
    [0.0, 0.5, 0.0, 0.5],
    [0.5, 0.0, 0.5, 0.0],
    [0.0, 0.5, 0.0, 0.5],
    [0.5, 0.0, 0.5, 0.0]
])

log_prob, secventa_stari = hmm.decode(
    observatii,
    algorithm="viterbi"
)

camere = secventa_stari + 1

probabilitate_comuna = np.exp(log_prob)

print("Secventa cea mai probabila:", camere)
print("Probabilitatea comuna:", probabilitate_comuna)
```
