```python
N = 10
total_apeluri = 180
rata_observata = total_apeluri / N


alpha = 2
beta = 0.1

media_prior = alpha / beta
dispersia_prior = alpha / beta**2

print("Media priorului:", media_prior)
print("Dispersia priorului:", dispersia_prior)

alpha_posterior = alpha + total_apeluri
beta_posterior = beta + N

print("Alpha posterior:", alpha_posterior)
print("Beta posterior:", beta_posterior)
print("Distributia posterioara: Gamma(", alpha_posterior, ",", beta_posterior, ")")

mod_posterior = (alpha_posterior - 1) / beta_posterior
print("Modulul posteriorului:", mod_posterior)

mostre = np.random.gamma(
    shape=alpha_posterior,
    scale=1 / beta_posterior,
    size=10000
)
interval_hdi = az.hdi(mostre, hdi_prob=0.94)

print("Intervalul HDI de 94%:", interval_hdi)
```
